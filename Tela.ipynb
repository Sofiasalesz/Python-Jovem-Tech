```python

import ipywidgets as widgets
from IPython.display import display, clear_output

# ==============================
# ESTOQUE FÁCIL
# ==============================

produtos = ["Teclado", "Mouse", "Monitor"]

# Cores e estilos
azul = "#2563EB"
azul_escuro = "#172554"
fundo = "#F1F5F9"
branco = "#FFFFFF"

# Título
titulo = widgets.HTML("""
<div style="
    background: linear-gradient(135deg, #172554, #2563EB);
    padding: 25px;
    border-radius: 15px;
    color: white;
    font-family: Arial;
">
    <h1 style="margin:0;">📦 ESTOQUE FÁCIL</h1>
    <p style="margin:8px 0 0 0;">
        Sistema de gerenciamento de produtos
    </p>
</div>
""")

# Campo de entrada
campo_produto = widgets.Text(
    placeholder="Digite o nome do produto...",
    description="Produto:",
    layout=widgets.Layout(width="100%"),
    style={"description_width": "initial"}
)

# Botões
btn_cadastrar = widgets.Button(
    description="＋ Cadastrar",
    button_style="success",
    layout=widgets.Layout(width="145px", height="40px")
)

btn_atualizar = widgets.Button(
    description="✏️ Atualizar",
    button_style="info",
    layout=widgets.Layout(width="145px", height="40px")
)

btn_excluir = widgets.Button(
    description="🗑️ Excluir",
    button_style="danger",
    layout=widgets.Layout(width="145px", height="40px")
)

btn_limpar = widgets.Button(
    description="Limpar campo",
    layout=widgets.Layout(width="145px", height="40px")
)

# Lista suspensa para selecionar produtos
selecao = widgets.Dropdown(
    options=produtos,
    description="Selecionar:",
    layout=widgets.Layout(width="100%"),
    style={"description_width": "initial"}
)

# Área de mensagens e tabela
mensagem = widgets.Output()
tabela = widgets.Output()

contador = widgets.HTML()

# Atualizar a interface
def atualizar_lista():
    selecao.options = produtos

    if produtos:
        if selecao.value not in produtos:
            selecao.value = produtos[0]

    contador.value = f"""
    <div style="
        background:#DBEAFE;
        color:#1D4ED8;
        padding:10px;
        border-radius:8px;
        font-weight:bold;
        margin:10px 0;
    ">
        📊 Total de produtos: {len(produtos)}
    </div>
    """

    with tabela:
        clear_output(wait=True)

        if produtos:
            linhas = ""
            for numero, produto in enumerate(produtos, start=1):
                linhas += f"""
                <tr>
                    <td style="padding:12px;border-bottom:1px solid #E2E8F0;">
                        {numero}
                    </td>
                    <td style="padding:12px;border-bottom:1px solid #E2E8F0;">
                        {produto}
                    </td>
                </tr>
                """

            display(widgets.HTML(f"""
            <table style="
                width:100%;
                border-collapse:collapse;
                background:white;
                font-family:Arial;
                text-align:left;
            ">
                <tr style="background:#DBEAFE;color:#172554;">
                    <th style="padding:12px;">Nº</th>
                    <th style="padding:12px;">Nome do produto</th>
                </tr>
                {linhas}
            </table>
            """))
        else:
            display(widgets.HTML(
                "<p>📭 Nenhum produto cadastrado.</p>"
            ))

# Exibir mensagens
def mostrar_mensagem(texto, cor="#2563EB"):
    with mensagem:
        clear_output(wait=True)
        display(widgets.HTML(f"""
        <div style="
            background:{cor}15;
            border-left:4px solid {cor};
            color:{cor};
            padding:12px;
            border-radius:6px;
            margin:10px 0;
            font-family:Arial;
        ">
            {texto}
        </div>
        """))

# Cadastrar
def cadastrar(b):
    nome = campo_produto.value.strip()

    if not nome:
        mostrar_mensagem("Digite o nome do produto!", "#DC2626")
        return

    produtos.append(nome)
    campo_produto.value = ""
    atualizar_lista()
    mostrar_mensagem("Produto cadastrado com sucesso!", "#15803D")

# Atualizar
def atualizar(b):
    antigo = selecao.value
    novo = campo_produto.value.strip()

    if antigo is None:
        mostrar_mensagem("Não há produtos para atualizar.", "#DC2626")
        return

    if not novo:
        mostrar_mensagem("Digite o novo nome no campo Produto.", "#DC2626")
        return

    indice = produtos.index(antigo)
    produtos[indice] = novo

    campo_produto.value = ""
    atualizar_lista()
    mostrar_mensagem("Produto atualizado com sucesso!", "#15803D")

# Excluir
def excluir(b):
    produto = selecao.value

    if produto is None:
        mostrar_mensagem("Não há produtos para excluir.", "#DC2626")
        return

    produtos.remove(produto)
    atualizar_lista()
    mostrar_mensagem("Produto excluído com sucesso!", "#15803D")

# Limpar campo
def limpar(b):
    campo_produto.value = ""
    mostrar_mensagem("Campo de texto limpo.")

# Conectar botões às funções
btn_cadastrar.on_click(cadastrar)
btn_atualizar.on_click(atualizar)
btn_excluir.on_click(excluir)
btn_limpar.on_click(limpar)

# Selecionar um produto preenche o campo para facilitar
def selecionar_produto(event):
    if event["new"] is not None:
        campo_produto.value = event["new"]

selecao.observe(selecionar_produto, names="value")

# Layout
botoes = widgets.HBox(
    [btn_cadastrar, btn_atualizar, btn_excluir, btn_limpar],
    layout=widgets.Layout(
        display="flex",
        flex_flow="row wrap",
        gap="8px"
    )
)

painel = widgets.VBox([
    titulo,
    widgets.HTML("<h3 style='color:#172554'>Gerenciar produtos</h3>"),
    campo_produto,
    selecao,
    botoes,
    mensagem,
    contador,
    widgets.HTML("<h3 style='color:#172554'>📋 Lista de produtos</h3>"),
    tabela
], layout=widgets.Layout(
    width="100%",
    max_width="850px",
    padding="15px",
    gap="10px"
))

atualizar_lista()
display(painel)
```
